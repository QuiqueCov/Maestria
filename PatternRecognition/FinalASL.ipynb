# Dividing

In this block is explained how I divided the data:
* 70 % of the information to train
* 15 % of the information to validate
* 15 % of the information to test

```python
train_size = int(len(data)*.7)
val_size = int(len(data)*.15) # before 20 % now is 15%
test_size = int(len(data)*.15) # before 10% now is 15%
```

```python
# sizes
print(train_size) ## 220 * 8 = 1760
print(val_size) ## 63 * 8 = 504
print(test_size) # 31 * 8 = 248
```

```python
type(data)
```

```python
train = data.take(train_size)
val = data.skip(train_size).take(val_size)
test = data.skip(train_size+val_size).take(test_size)
```

```python
fig, ax = plt.subplots(ncols=8, figsize=(20,20))
train_iterator = train.as_numpy_iterator()
train_batch = train_iterator.next()
for idx, img in enumerate(train_batch[0][:10]):
    ax[idx].imshow(img.astype(int))
    ax[idx].title.set_text(train_batch[1][idx])  # labels
plt.show()
```

#Model Definition


```python
i =Input(shape=(128,128,1)) # 128x128, 1 = gray_scale
# First Layer
x = Conv2D(32, (2,2), activation='relu', padding='same')(i) # Padding same, in order to dont modify the image size
x = BatchNormalization()(x)
x = Conv2D(32, (2,2), activation = 'relu', padding='same')(x)
x = BatchNormalization()(x)
x = MaxPooling2D((2,2))(x)

# Second Layer
x = Conv2D(64, (2,2), activation='relu', padding='same')(x) # Padding same, in order to dont modify the image size
x = BatchNormalization()(x)
x = Conv2D(64, (2,2), activation = 'relu', padding='same')(x)
x = BatchNormalization()(x)
x = MaxPooling2D((2,2))(x)

# Third Layer
x = Conv2D(128, (2,2), activation='relu', padding='same')(x) # Padding same, in order to dont modify the image size
x = BatchNormalization()(x)
x = Conv2D(128, (2,2), activation = 'relu', padding='same')(x)
x = BatchNormalization()(x)
x = MaxPooling2D((2,2))(x)

# Fourth Layer
x = Conv2D(256, (2,2), activation='relu', padding='same')(x) # Padding same, in order to dont modify the image size
x = BatchNormalization()(x)
x = Conv2D(256, (2,2), activation = 'relu', padding='same')(x)
x = BatchNormalization()(x)
x = MaxPooling2D((2,2))(x)


x = Flatten()(x)
x = Dropout(0.2)(x)
x = Dense(512, activation='relu')(x)
x = Dropout(0.2)(x)
x = Dense(K, activation='softmax')(x)

model = Model(i, x)
```

```python
model
```

```python
# compile
model.compile(optimizer = 'adam',
              loss = 'sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
r = model.fit(train, validation_data = val, epochs=50)
```

```python
plt.plot(r.history['loss'], label = 'loss')
plt.plot(r.history['val_loss'], label = 'val_loss')
plt.legend()
```

```python
plt.plot(r.history['accuracy'], label='acc')
plt.plot(r.history['val_accuracy'], label='val_acc')
plt.legend()
```

```python
model.summary()
```

# Make Predictions

```python
test_iterator = test.as_numpy_iterator() # using a test iterator

for batch in test_iterator: # iteraring throught the batches
    images, true_labels = batch
    batch_predictions = model.predict(images) # getting the prediction pool
    batch_predicted_labels = np.argmax(batch_predictions, axis=1) # getting the predicted label

    fig, axes = plt.subplots(nrows=1, ncols=len(images), figsize=(20, 5)) # plotting the images
    for i in range(len(images)):
      axes[i].imshow(images[i].astype(int))
      axes[i].set_title(f"Pred:{data_names[batch_predicted_labels[i]]}\n True:{data_names[true_labels[i]]}") # predicted vs true label
      axes[i].axis('off')
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import confusion_matrix

all_true_labels = []
all_predicted_labels = []

atest_iterator = test.as_numpy_iterator()

for batch in atest_iterator:
    images, true_labels = batch
    batch_predictions = model.predict(images)
    batch_predicted_labels = np.argmax(batch_predictions, axis=1)

    all_true_labels.extend(true_labels)
    all_predicted_labels.extend(batch_predicted_labels)

conf_matrix = confusion_matrix(all_true_labels, all_predicted_labels)


print(conf_matrix)


import seaborn as sns

plt.figure(figsize=(10, 8))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=data_names, yticklabels=data_names)
plt.xlabel('Etiquetas Predichas')
plt.ylabel('Etiquetas Verdaderas')
plt.title('Matriz de Confusión')
plt.show()
```

```python
import numpy as np
from sklearn.metrics import classification_report

all_true_labels = []
all_predicted_labels = []

test_iterator = test.as_numpy_iterator() # using a test iterator

for batch in test_iterator: # iteraring throught the batches
    images, true_labels = batch
    batch_predictions = model.predict(images) # getting the prediction pool
    batch_predicted_labels = np.argmax(batch_predictions, axis=1) # getting the predicted label

    # Acumula las etiquetas en las listas
    all_true_labels.extend(true_labels)
    all_predicted_labels.extend(batch_predicted_labels)

# Genera el reporte de clasificación
print(classification_report(all_true_labels, all_predicted_labels, target_names=data_names))
```
